## gabor only

```python
import matplotlib.pyplot as plt
import numpy as np
import torch

data = np.load(
    "dwt_checkpoints/kodak/GaussianImage_Cholesky_150000_5000/kodim01/training.npy",
    allow_pickle=True
).item()

iterations = np.array(data['iterations'])
train_psnr_curve = np.array(data['training_psnr'])

fig, ax1 = plt.subplots()

# PSNR
ax1.plot(iterations, train_psnr_curve, label="PSNR")
ax1.set_xlabel("Iteration")
ax1.set_ylabel("PSNR (dB)")
loss = np.array([l.cpu().item() for l in data['loss']])
ax2 = ax1.twinx()
ax2.plot(iterations, loss, linestyle="--", label="Loss")
ax2.set_ylabel("Loss")

plt.title("Training Curve")
ax1.grid(True)

plt.show()
```
